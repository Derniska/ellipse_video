```python
import socket
import requests

IP = "83.172.1.195"
PORT = 673

print(f"--- 1. Проверка доступности TCP-порта {PORT} ---")
sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
sock.settimeout(3.0)

try:
    result = sock.connect_ex((IP, PORT))
    if result == 0:
        print(f" УСПЕХ: Порт {PORT} на {IP} открыт и доступен из сети.")
    else:
        print(f" ОШИБКА: Порт {PORT} закрыт или заблокирован (код: {result}).")
        print("  -> Проверьте проброс портов (Port Forwarding) на роутере в магазине.")
finally:
    sock.close()

print(f"\n--- 2. Проверка HTTP-ответа ---")
url_http = f"http://{IP}:{PORT}/"

try:
    response = requests.get(url_http, timeout=4)
    print(f" HTTP-ответ получен! Код статуса: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f" Запрос по HTTP не прошел: {e}")
```
